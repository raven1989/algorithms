# 110.平衡二叉树

[力扣题目链接](https://leetcode.cn/problems/balanced-binary-tree/)

给定一个二叉树，判断它是否是高度平衡的二叉树。

本题中，一棵高度平衡二叉树定义为：一个二叉树每个节点 的左右两个子树的高度差的绝对值不超过1。

示例 1:

给定二叉树 [3,9,20,null,null,15,7]

![110.平衡二叉树](https://file1.kamacoder.com/i/algo/2021020315542230.png)

返回 true 。

示例 2:

给定二叉树 [1,2,2,3,3,null,null,4,4]

![110.平衡二叉树1](https://file1.kamacoder.com/i/algo/20210203155447919.png)

返回 false 。


```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Tree:
    @staticmethod
    def build(arr):
        q = []
        i = j = 0
        while j < len(arr):
            if len(q) == 0:
                q.append(TreeNode(arr[j]))
                j += 1
            else:
                q[i].left = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                q[i].right = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                if q[i].left:
                    q.append(q[i].left)
                if q[i].right:
                    q.append(q[i].right)
                i += 1
        # print([it.val if it else it for it in q])
        return q[0] if len(q)>0 else None
    
    @staticmethod
    def print(root):
        ## 如何print一棵树，如果左和右子树都画好了，就能画根了
        def _postorder_print(r):
            if not r:
                return ['']
            left = _postorder_print(r.left)
            right = _postorder_print(r.right)
            res = ['']
            lines = max(len(left), len(right))
            for i in range(lines):
                ll = left[i] if i<len(left) else ' '*len(left[0])
                rr = right[i] if i<len(right) else ' '*len(right[0])
                res.append([ll, ' '*len(str(r.val)), rr])
            res[0] = [
                '' if len(res)<2 else ' '*len(res[1][0]),
                str(r.val),
                '' if len(res)<2 else ' '*len(res[1][2])
            ]
            res = [''.join(line) for line in res]
            # print(res)
            return res
        return '\n'.join([line for line in _postorder_print(root) if line.strip()!=''])
```

```python
def isBalanced(root):

    def height(r):
        if not r:
            return 0
        h1 = height(r.left)
        h2 = height(r.right)
        ## 如果子树中已经有imbalanced，不再传递高度，返回标记-1
        if h1 < 0 or h2 < 0:
            return -1
        if abs(h1-h2) > 1:
            return -1
        return 1 + max(h1,h2)
    return height(root) >= 0
```

```python
root = Tree.build([3,9,20,None,None,15,7])
print("Tree:")
print(Tree.print(root))
isBalanced(root)
```

```python
root = Tree.build([1,2,2,3,3,None,None,4,4])
print("Tree:")
print(Tree.print(root))
isBalanced(root)
```

```python
root = Tree.build([])
print("Tree:")
print(Tree.print(root))
isBalanced(root)
```
