# 53. 最大子序和

[力扣题目链接](https://leetcode.cn/problems/maximum-subarray/)

给定一个整数数组 nums ，找到一个具有最大和的连续子数组（子数组最少包含一个元素），返回其最大和。

示例:

- 输入: [-2,1,-3,4,-1,2,1,-5,4]
- 输出: 6
- 解释:  连续子数组  [4,-1,2,1] 的和最大，为  6。


```python
def maxSubArray(nums):
    return greedy(nums)

def greedy(nums):
    ## curr sum如果小于0，就从下一个开始重新累计，因为无论下一个是正或负，加上前面的都是更小
    ## curr sum如果大于0，加上下一个一定更大，而左边第一个数一定是非负数开始的，因为第一条
    if len(nums) == 0:
        return 0
    max_sum, curr_sum = nums[0], 0
    for x in nums:
        curr_sum += x
        if curr_sum > max_sum:
            max_sum = curr_sum
        if curr_sum < 0:
            curr_sum = 0
    return max_sum
```

```python
nums = [-2,1,-3,4,-1,2,1,-5,4]
maxSubArray(nums)
```

```python
nums = [1]
maxSubArray(nums)
```

```python
nums = [5,4,-1,7,8]
maxSubArray(nums)
```
