# 416. 分割等和子集

[力扣题目链接](https://leetcode.cn/problems/partition-equal-subset-sum/)

题目难易：中等

给定一个只包含正整数的非空数组。是否可以将这个数组分割成两个子集，使得两个子集的元素和相等。

注意:
每个数组中的元素不会超过 100
数组的大小不会超过 200

示例 1:
* 输入: [1, 5, 11, 5]
* 输出: true
* 解释: 数组可以分割成 [1, 5, 5] 和 [11].

示例 2:
* 输入: [1, 2, 3, 5]
* 输出: false
* 解释: 数组不能分割成两个元素和相等的子集.

提示：
* 1 <= nums.length <= 200
* 1 <= nums[i] <= 100

```python
def canPartition(nums):
    ## 等价于能否填满sum/2的01背包问题：
    ## dp[i][j]表示从0-i中任选items能否填满j
    capacity = sum(nums)
    if capacity % 2 == 1:
        return False
    capacity = capacity // 2
    dp = [0] * (capacity + 1)
    dp[0] = 1

    ## 如果不取item i，那么能不能填满j就=dp[i-1][j]，一位数组下就是dp[j]
    ## 如果取item i，那么就得看dp[i-1][j-value[j]]能不能填满
    ## 注意，要使用i-1来更新，所以一维数组的话，要从后往前更新
    for i in range(len(nums)):
        for j in range(capacity, nums[i]-1, -1):
            dp[j] = max(dp[j], dp[j-nums[i]])
        # print(dp)
    return dp[-1]==1
```

```python
nums = [1,5,11,5]
canPartition(nums)
```

```python
nums = [1,2,3,5]
canPartition(nums)
```
