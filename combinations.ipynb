# 第77题. 组合

[力扣题目链接](https://leetcode.cn/problems/combinations/ )

给定两个整数 n 和 k，返回 1 ... n 中所有可能的 k 个数的组合。

示例:
输入: n = 4, k = 2
输出:
[
  [2,4],
  [3,4],
  [2,3],
  [1,2],
  [1,3],
  [1,4],
]

```python
def conbine(n, k):
    res = []
    path = []

    def backtrack(path, n, start, k, res):
        if len(path) == k:
            # print(path)
            res.append([x for x in path])
            return
        for i in range(start, n+1):
            path.append(i)
            backtrack(path, n, i+1, k, res)
            path.pop()
    
    ## 带剪枝的回溯
    def backtrack_trim(path, n, start, k, res):
        if len(path) == k:
            res.append([x for x in path])
            return
        end = (n+1) - (k - len(path)) + 1
        for i in range(start, end):
            path.append(i)
            backtrack(path, n, i+1, k, res)
            path.pop()

    backtrack_trim(path, n, 1, k, res)
    return res
```

```python
conbine(4, 2)
```

```python
conbine(1, 1)
```
