# 518.零钱兑换II

[力扣题目链接](https://leetcode.cn/problems/coin-change-ii/)

给定不同面额的硬币和一个总金额。写出函数来计算可以凑成总金额的硬币组合数。假设每一种面额的硬币有无限个。 

示例 1:

* 输入: amount = 5, coins = [1, 2, 5]
* 输出: 4

解释: 有四种方式可以凑成总金额:

* 5=5
* 5=2+2+1
* 5=2+1+1+1
* 5=1+1+1+1+1

示例 2:

* 输入: amount = 3, coins = [2]
* 输出: 0
* 解释: 只用面额2的硬币不能凑成总金额3。

示例 3:
* 输入: amount = 10, coins = [10]
* 输出: 1

注意，你可以假设：

* 0 <= amount (总金额) <= 5000
* 1 <= coin (硬币面额) <= 5000
* 硬币种类不超过 500 种
* 结果符合 32 位符号整数

```python
def change2(amount, coins):
    ## 完全背包问题，因为coins可以重复选
    ## dp[i][j]表示从0-i中任选，amount为j的组合数
    ## 遍历顺序决定是组合还是排列：
    ## 如果先遍历物品再背包容量就是组合：因为coins本身是有顺序的，不会出现21这个顺序的组合
    ## 如果先遍历背包容量再物品就是排列：思考下含义是背包容量为j的情况下去遍历coins的0-i，所以是排列
    dp = [0] * (amount+1)
    dp[0] = 1

    ## 如果完全不放coins[i]，dp[i][j]则是dp[i-1][j]
    ## 如果放一个coins[i]，dp[i][j]则是dp[i][j-coins[i]]，
    ## 不是dp[i-1][j-coins[i]]的原因是amount为j-coins[i]时，也是可以放coins[i]的
    for i in range(len(coins)):
        for j in range(coins[i], amount+1):
            dp[j] = dp[j] + dp[j-coins[i]] ## 没有+1 or 其他，所以dp[0]一定不能初始化为0，负责如何递推都是0
        # print(dp)
    return dp[-1]
```

```python
amount = 5
coins = [1,2,5]
change2(amount, coins)
```

```python
amount = 3
coins = [2]
change2(amount, coins) 
```

```python
amount = 10
coins = [10]
change2(amount, coins) 
```

# 322. 零钱兑换

[力扣题目链接](https://leetcode.cn/problems/coin-change/)

给定不同面额的硬币 coins 和一个总金额 amount。编写一个函数来计算可以凑成总金额所需的最少的硬币个数。如果没有任何一种硬币组合能组成总金额，返回 -1。

你可以认为每种硬币的数量是无限的。

示例 1：
* 输入：coins = [1, 2, 5], amount = 11
* 输出：3
* 解释：11 = 5 + 5 + 1

示例 2：
* 输入：coins = [2], amount = 3
* 输出：-1

示例 3：
* 输入：coins = [1], amount = 0
* 输出：0

示例 4：
* 输入：coins = [1], amount = 1
* 输出：1

示例 5：
* 输入：coins = [1], amount = 2
* 输出：2

提示：

* 1 <= coins.length <= 12
* 1 <= coins[i] <= 2^31 - 1
* 0 <= amount <= 10^4


```python
def change(coins, amount):
    ## 也是完全背包问题
    ## dp[i][j]表示从0-i中任选，amount为j的组合数
    dp = [float('inf')] * (amount+1)
    dp[0] = 0
    # print(dp)

    ## 如果完全不放coins[i]，dp[i][j]则是dp[i-1][j]
    ## 如果放一个coins[i]，dp[i][j]则是dp[i][j-coins[i]]+1，
    ## dp[i][j] = min(不选i，当前选一个i)
    ## 不是dp[i-1][j-coins[i]]的原因是amount为j-coins[i]时，也是可以放coins[i]的
    for i in range(len(coins)):
        for j in range(coins[i], amount+1):
            dp[j] = min(dp[j], dp[j-coins[i]]+1)
        # print(dp)
    return dp[-1] if dp[-1]<float('inf') else -1
```

```python
coins = [1,2,5]
amount = 11
change(coins, amount)
```

```python
coins = [2]
amount = 3
change(coins, amount)
```

```python
coins = [1]
amount = 0
change(coins, amount)
```
