# 20. 有效的括号

[力扣题目链接](https://leetcode.cn/problems/valid-parentheses/)

给定一个只包括 '('，')'，'{'，'}'，'['，']' 的字符串，判断字符串是否有效。

有效字符串需满足：
* 左括号必须用相同类型的右括号闭合。
* 左括号必须以正确的顺序闭合。
* 注意空字符串可被认为是有效字符串。

示例 1:
* 输入: "()"  
* 输出: true     

示例 2:
* 输入: "()[]{}"    
* 输出: true     

示例 3:
* 输入: "(]"   
* 输出: false    

示例 4:
* 输入: "([)]"   
* 输出: false 

示例 5:
* 输入: "{[]}"
* 输出: true


```python
def isValid(s):
    d = {')':'(', ']':'[', '}':'{'}
    stack = []
    for x in s:
        if x in d:
            if len(stack)==0 or stack.pop()!=d[x]:
                return False
        else:
            stack.append(x)
    return len(stack) == 0
```

```python
s = "()"
isValid(s)
```

```python
s = "()[]{}"
isValid(s)
```

```python
s = "(]"
isValid(s)
```

```python
s = "([])"
isValid(s)
```

```python
s = "([)]"
isValid(s)
```

```python
s = ""
isValid(s)
```
