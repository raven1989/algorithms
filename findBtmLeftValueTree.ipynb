给定一个二叉树，在树的最后一行找到最左边的值。

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Tree:
    @staticmethod
    def build(arr):
        q = []
        i = j = 0
        while j < len(arr):
            if len(q) == 0:
                q.append(TreeNode(arr[j]))
                j += 1
            else:
                q[i].left = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                q[i].right = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                if q[i].left:
                    q.append(q[i].left)
                if q[i].right:
                    q.append(q[i].right)
                i += 1
        # print([it.val if it else it for it in q])
        return q[0] if len(q)>0 else None
    
    @staticmethod
    def print(root):
        ## 如何print一棵树，如果左和右子树都画好了，就能画根了
        def _postorder_print(r):
            if not r:
                return ['']
            left = _postorder_print(r.left)
            right = _postorder_print(r.right)
            res = ['']
            lines = max(len(left), len(right))
            for i in range(lines):
                ll = left[i] if i<len(left) else ' '*len(left[0])
                rr = right[i] if i<len(right) else ' '*len(right[0])
                res.append([ll, ' '*len(str(r.val)), rr])
            res[0] = [
                '' if len(res)<2 else ' '*len(res[1][0]),
                str(r.val),
                '' if len(res)<2 else ' '*len(res[1][2])
            ]
            res = [''.join(line) for line in res]
            # print(res)
            return res
        return '\n'.join([line for line in _postorder_print(root) if line.strip()!=''])
```

```python
def findBottomLeftValue(root):
    return bfs(root)

def bfs(root):
    if not root:
        return root
    visit = root
    ## True表示是左子
    q = [root]
    while len(q) > 0:
        visit = q[0]
        next_level = []
        for n in q:
            if n.left:
                next_level.append(n.left)
            if n.right:
                next_level.append(n.right)
        q = next_level
    return visit.val
```

```python
root = Tree.build([2,1,3])
print("Tree:")
print(Tree.print(root))
findBottomLeftValue(root)
```

```python
root = Tree.build([1,2,3,4,None,5,6,None,None,7])
print("Tree:")
print(Tree.print(root))
findBottomLeftValue(root)
```

```python
root = Tree.build([0,None,-1])
print("Tree:")
print(Tree.print(root))
findBottomLeftValue(root)
```
