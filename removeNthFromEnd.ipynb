# 19.删除链表的倒数第N个节点

[力扣题目链接](https://leetcode.cn/problems/remove-nth-node-from-end-of-list/)

给你一个链表，删除链表的倒数第 n 个结点，并且返回链表的头结点。

进阶：你能尝试使用一趟扫描实现吗？

示例 1：

输入：head = [1,2,3,4,5], n = 2
输出：[1,2,3,5]

示例 2：

输入：head = [1], n = 1
输出：[]

示例 3：

输入：head = [1,2], n = 1
输出：[1]

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def to_array(self):
        arr = []
        curr = self
        while curr:
            arr.append(curr.val)
            curr = curr.next
        return arr

def removeNthFromEnd(head, n):
    # return removeNthFromEnd_loop(head, n)
    _, head = removeNthFromEnd_recursive(head, n)
    return head

def removeNthFromEnd_loop(head, n):
    handle = ListNode(-1, head)
    slow = fast = handle
    for _ in range(n):
        if fast:
            fast = fast.next
    if not fast:
        head = handle.next
        del handle
        return head
    
    while fast.next:
        slow = slow.next
        fast = fast.next
    
    discard = slow.next
    slow.next = slow.next.next
    del discard

    head = handle.next
    del handle
    return head

def removeNthFromEnd_recursive(head, n):
    if not head:
        return 1, head
    k, node = removeNthFromEnd_recursive(head.next, n)
    head.next = node
    if k == n:
        rest = head.next
        del head
        return k+1, rest
    else:
        return k+1, head
```

```python
def make_list(arr):
    arr.reverse()
    head = None
    for x in arr:
        head = ListNode(x, head)
    return head
```

```python
head = make_list([1,2,3,4,5])
print(head.to_array() if head else [])
head = removeNthFromEnd(head, 2)
print(head.to_array() if head else [])
```

```python
head = make_list([1])
print(head.to_array() if head else [])
head = removeNthFromEnd(head, 1)
print(head.to_array() if head else [])
```
