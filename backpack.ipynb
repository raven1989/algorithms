# 01 背包

有n件物品和一个最多能背重量为w 的背包。第i件物品的重量是weight[i]，得到的价值是value[i] 。**每件物品只能用一次**，求解将哪些物品装入背包里物品价值总和最大。

背包最大重量为4。

物品为：

|       | 重量 | 价值 |
| ----- | ---- | ---- |
| 物品0 | 1    | 15   |
| 物品1 | 3    | 20   |
| 物品2 | 4    | 30   |

问背包能背的物品最大价值是多少？

以下讲解和图示中出现的数字都是以这个例子为例。

```python
def getMaxValue(weights, values, capacity):
    return dp_2dim(weights, values, capacity)
    # return dp_1dim(weights, values, capacity)

def dp_2dim(weights, values, capacity):
    ## dp[i][j]的含义是从0-i中任意选择物品满足背包容量<=j的条件下的最大value
    dp = [[0 for _ in range(capacity+1)] for _ in range(len(weights))]

    ## init dp[0]
    for j in range(weights[0], capacity+1):
        dp[0][j] = values[0]
    print(dp)

    for i in range(1, len(weights)):
        for j in range(0, capacity+1):
            ## 更新公式：
            ## 如果j<weights[i]，不能选i，最大值就是dp[i-1][j]
            ## 如果可以选i：
            ## 不选i的话，capacity为j的最大值就是dp[i-1][j]
            ## 选i的话，capacity为j的最大值就是values[i] + dp[i-1][j-values[i]]
            if j < weights[i]:
                dp[i][j] = dp[i-1][j]
            else:
                dp[i][j] = max(
                    dp[i-1][j], 
                    values[i] + dp[i-1][j-weights[i]]
                )
    print(dp)
    return dp[-1][-1]

def dp_1dim(weights, values, capacity):
    ## dp[i][j]的含义是从0-i中任意选择物品满足背包容量<=j的条件下的最大value
    ## 但可以只用一行数组节省空间
    dp = [0] * (capacity+1)

    ## init dp[0] = 0

    ## 这次从0行开始，兼容初始化
    for i in range(len(weights)):
        for j in range(capacity, weights[i]-1, -1):
            ## 更新公式：
            ## 如果j<weights[i]，不能选i，最大值就是dp[i-1][j]，就是dp[j]，所以就循环到weights[i]
            ## 如果可以选i：
            ## 不选i的话，capacity为j的最大值就是dp[i-1][j]
            ## 选i的话，capacity为j的最大值就是values[i] + dp[i-1][j-values[i]]
            dp[j] = max(dp[j], values[i] + dp[j-weights[i]])
    return dp[-1]
```

```python
weights = [1,3,4]
values = [15,20,30]
capacity = 4
getMaxValue(weights, values, capacity)
```

```python
weights = [2,2,3,1,5,2]
values = [2,3,1,5,4,3]
capacity = 1
getMaxValue(weights, values, capacity)
```

## 完全背包

有N件物品和一个最多能背重量为W的背包。第i件物品的重量是weight[i]，得到的价值是value[i] 。**每件物品都有无限个（也就是可以放入背包多次）**，求解将哪些物品装入背包里物品价值总和最大。

**完全背包和01背包问题唯一不同的地方就是，每种物品有无限件**。

同样leetcode上没有纯完全背包问题，都是需要完全背包的各种应用，需要转化成完全背包问题，所以我这里还是以纯完全背包问题进行讲解理论和原理。

在下面的讲解中，我拿下面数据举例子：

背包最大重量为4，物品为：

|       | 重量 | 价值 |
| ----- | ---- | ---- |
| 物品0 | 1    | 15   |
| 物品1 | 3    | 20   |
| 物品2 | 4    | 30   |

**每件商品都有无限个！**

问背包能背的物品最大价值是多少？


```python
def getMaxValue2(weights, values, capacity):
    # return dp_2dim2(weights, values, capacity)
    return dp_1dim2(weights, values, capacity)

def dp_2dim2(weights, values, capacity):
    ## dp[i][j]还是表示从0-i中任选，重量<=j的最大value
    ## 如果不选i，则dp[i][j]就是dp[i-1][j]
    ## 如果选i，则dp[i][j]就是dp[i][j-weights[i]]+values[i]，
    ## 这里体现了完全背包与01背包的区别：物品i可以多次选择，所以不是dp[i-1][j-weights[i]]
    ## 初始值：dp[.][0] = 0 表示容量为0的价值为0，注意前提是0号物品的weight不为0
    dp = [[0]*(capacity+1) for _ in range(len(weights))]

    for i in range(len(weights)):
        for j in range(capacity+1):
            if j < weights[i]:
                dp[i][j] = dp[i-1][j]
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-weights[i]] + values[i])
        # print(dp[i])
    return dp[-1][-1]

def dp_1dim2(weights, values, capacity):
    dp = [0] * (capacity+1)

    for i in range(len(weights)):
        ## 由于递推公式中会用到dp[i]，所以j不能像01背包那样倒着更新了
        for j in range(weights[i], capacity+1):
            dp[j] = max(dp[j], dp[j-weights[i]]+values[i])
        # print(dp)
    return dp[-1]
```

```python
weights = [1,3,4]
values = [15,20,30]
capacity = 4
getMaxValue2(weights, values, capacity)
```
