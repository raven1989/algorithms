# 42. 接雨水

[力扣题目链接](https://leetcode.cn/problems/trapping-rain-water/)

给定 n 个非负整数表示每个宽度为 1 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

示例 1：

![image](https://file1.kamacoder.com/i/algo/2025-07-04_12-06-59.jpg) 
* 输入：height = [0,1,0,2,1,0,1,3,2,1,2,1]
* 输出：6
* 解释：上面是由数组 [0,1,0,2,1,0,1,3,2,1,2,1] 表示的高度图，在这种情况下，可以接 6 个单位的雨水（蓝色部分表示雨水）。

示例 2：

* 输入：height = [4,2,0,3,2,5]
* 输出：9


```python
def tap(height):
    return tap2(height)

def tap1(height):
    ## 记录每个位置的左和右的最高height
    left_max = [0] * len(height)
    right_max = [0] * len(height)

    left_max[0], right_max[-1] = height[0], height[-1]
    for i in range(1, len(height)):
        left_max[i] = max(left_max[i-1], height[i])
        right_max[-i-1] = max(right_max[-i],height[-i-1])

    water = 0
    for i in range(1, len(height)-1):
        water += (min(left_max[i], right_max[i]) - height[i])
    return water

def tap2(height):
    water = 0
    ## 用双指针动态记录左边界和右边界
    l, r = 0, len(height)-1
    ## 再记录l_max和r_max，分别表示从左右遍历过程中的最高的height
    l_max, r_max = height[0], height[-1]
    ## 如何遍历？左右谁小就移动谁，对当前移动了的l和r，计算能存多少水
    while l < r:
        if height[l] <= height[r]:
            l += 1
            l_max = max(l_max, height[l])
            water += max(0, min(l_max, r_max) - height[l])
        else:
            r -= 1
            r_max = max(r_max, height[r])
            water += max(0, min(l_max, r_max) - height[r])
        # print(height[l_max:r_max+1], l, r, water)
    return water
```

```python
height = [0,1,0,2,1,0,1,3,2,1,2,1]
tap(height)
```
