# 203.移除链表元素

[力扣题目链接](https://leetcode.cn/problems/remove-linked-list-elements/)

题意：删除链表中等于给定值 val 的所有节点。

示例 1：
输入：head = [1,2,6,3,4,5,6], val = 6
输出：[1,2,3,4,5]

示例 2：
输入：head = [], val = 1
输出：[]

示例 3：
输入：head = [7,7,7,7], val = 7
输出：[]

```python
#Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def to_array(self):
        arr = []
        curr = self
        while curr:
            arr.append(curr.val)
            curr = curr.next
        return arr
```

```python
def removeFromLink_loop(head, val):
    handle = ListNode(0, head)
    prev, curr = handle, head
    while curr:
        if curr.val == val:
            prev.next = curr.next
        else:
            prev = curr
        curr = curr.next
    return handle.next

def removeFromLink_recursive(head, val):
    if not head:
        return None
    head.next = removeFromLink_recursive(head.next, val)
    return head.next if head.val == val else head
```

```python
def removeFromLink(head, val):
    # return removeFromLink_loop(head, val)
    return removeFromLink_recursive(head, val)
```

```python
head = ListNode(1, ListNode(2, ListNode(6, ListNode(3, ListNode(4, ListNode(5, ListNode(6)))))))
head = removeFromLink(head, 6)
print(head.to_array())
```

```python
head = None
head = removeFromLink(head, 1)
print(head.to_array() if head else None)
```

```python
head = ListNode(7, ListNode(7, ListNode(7, ListNode(7))))
head = removeFromLink(head, 7)
print(head.to_array() if head else None)
```
