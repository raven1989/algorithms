# 142.环形链表II

[力扣题目链接](https://leetcode.cn/problems/linked-list-cycle-ii/)

题意：
给定一个链表，返回链表开始入环的第一个节点。 如果链表无环，则返回 null。

为了表示给定链表中的环，使用整数 pos 来表示链表尾连接到链表中的位置（索引从 0 开始）。 如果 pos 是 -1，则在该链表中没有环。

**说明**：不允许修改给定的链表。

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def to_array(self, n):
        arr = []
        curr = self
        i = 0
        while curr and i<n:
            arr.append(curr.val)
            curr = curr.next
            i += 1
        return arr
def detectCycle(head):
    slow = fast = head
    while slow and fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            p = head
            while p is not slow:
                p = p.next
                slow = slow.next
            return p
    return None
```

```python
def make_list(arr, pos):
    arr.reverse()
    head = None
    tail = None
    p = None
    for i,x in enumerate(arr):
        head = ListNode(x, head)
        if i==0:
            tail = head
        if i==len(arr) - 1 - pos:
            p = head
    if tail and pos>0:
        tail.next = p
    return head
```

```python
vals, pos = [3,2,0,-4], 1
head = make_list(vals, pos)
print(head.to_array(len(vals) * 2 - pos) if head else [])
entrance = detectCycle(head)
print(entrance.val if entrance else entrance)
```

```python
vals, pos = [1, 2], 0
head = make_list(vals, pos)
print(head.to_array(len(vals) * 2 - pos) if head else [])
entrance = detectCycle(head)
print(entrance.val if entrance else entrance)
```

```python
vals, pos = [], 0
head = make_list(vals, pos)
print(head.to_array(len(vals) * 2 - pos) if head else [])
entrance = detectCycle(head)
print(entrance.val if entrance else entrance)
```

```python
vals, pos = [1, 2], -1
head = make_list(vals, pos)
print(head.to_array(len(vals) * 2 - pos) if head else [])
entrance = detectCycle(head)
print(entrance.val if entrance else entrance)
```
