# 59.螺旋矩阵II

[力扣题目链接](https://leetcode.cn/problems/spiral-matrix-ii/)

给定一个正整数 n，生成一个包含 1 到 n^2 所有元素，且元素按顺时针顺序螺旋排列的正方形矩阵。

示例:

输入: 3
输出:
[
 [ 1, 2, 3 ],
 [ 8, 9, 4 ],
 [ 7, 6, 5 ]
]



```python
def spiral_matrix(n):
    ma = [[0] * n for _ in range(n)]
    e = 1
    cnt, size = 0, n * n
    h, t = 0, n - 1
    while cnt < size:
        if h == t:
            ma[h][h] = e
            e += 1
            cnt += 1
        for i in range(h, t):
            ma[h][i] = e
            e += 1
            cnt += 1
        for i in range(h, t):
            ma[i][t] = e
            e += 1
            cnt += 1
        for i in range(t, h, -1):
            ma[t][i] = e
            e += 1
            cnt += 1
        for i in range(t, h, -1):
            ma[i][h] = e
            e += 1
            cnt += 1
        h += 1
        t -= 1
    return ma
```

```python
spiral_matrix(1)
```
