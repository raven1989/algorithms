# 215 Kth Largest Element in Array

Given an integer array nums and an integer k, return the kth largest element in the array.

Note that it is the kth largest element in the sorted order, not the kth distinct element.

Can you solve it without sorting?

 

Example 1:

Input: nums = [3,2,1,5,6,4], k = 2

Output: 5

Example 2:

Input: nums = [3,2,3,1,2,4,5,5,6], k = 4

Output: 4
 

Constraints:

1 <= k <= nums.length <= 105
-104 <= nums[i] <= 104

```python
def quickSort(nums, left, right):
    if left >= right:
        return
    l, r = left, right
    v = nums[r]
    while l < r:
        while l < r and nums[l] > v:
            l += 1
        nums[r] = nums[l]
        while l < r and nums[r] <= v:
            r -= 1
        nums[l] = nums[r]
    nums[l] = v
    return quickSort(nums, left, l - 1), quickSort(nums, l + 1, right)

def findKthLargest_recursive(nums, k):
    def _findk(nums, left, right, k):
        if left >= right:
            return nums[left]
        l, r = left, right
        v = nums[r]
        while l < r:
            while l < r and nums[l] > v:
                l += 1
            nums[r] = nums[l]
            while l < r and nums[r] <= v:
                r -= 1
            nums[l] = nums[r]
        nums[l] = v
        if l == k-1:
            return nums[l]
        elif l < k:
            return _findk(nums, l + 1, right, k)
        else:
            return _findk(nums, left, l - 1, k)
    return _findk(nums, 0, len(nums) - 1, k)

def findKthLargest_loop(nums, k):
    left, right = 0, len(nums) - 1
    while left < right:
        l, r = left, right
        v = nums[r]
        while l < r:
            while l < r and nums[l] > v:
                l += 1
            nums[r] = nums[l]
            while l < r and nums[r] <= v:
                r -= 1
            nums[l] = nums[r]
        nums[l] = v
        # print(nums[left:right+1])
        if l == k - 1:
            return nums[l]
        elif l < k - 1:
            left = l + 1
        else:
            right = l - 1
    return nums[left]

def findKthLargest(nums, k):
    # exchange space for time
    left = [x for x in nums if x > nums[k-1]]
    right = [x for x in nums if x < nums[k-1]]
    middle = [x for x in nums if x == nums[k-1]]

    if k <= len(left):
        return findKthLargest(left, k)
    elif k > len(left) + len(middle):
        return findKthLargest(right, k - len(left) - len(middle))
    else:
        return middle[0]
```

```python
nums = [3,2,3,1,2,4,5,5,6]
quickSort(nums, 0, len(nums) - 1)
print(nums)
```

```python
nums = [3,2,3,1,2,4,5,5,6]
print(findKthLargest(nums, 4))
nums = [3,2,1,5,6,4]
print(findKthLargest(nums, 2))
nums = [1]
print(findKthLargest(nums, 1))
```
