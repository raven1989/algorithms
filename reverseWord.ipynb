# 151.翻转字符串里的单词

[力扣题目链接](https://leetcode.cn/problems/reverse-words-in-a-string/)

给定一个字符串，逐个翻转字符串中的每个单词。

示例 1：  
输入: "the sky is blue"   
输出: "blue is sky the"    

示例 2：   
输入: "  hello world!  "   
输出: "world! hello"      
解释: 输入字符串可以在前面或者后面包含多余的空格，但是反转后的字符不能包括。 

示例 3：   
输入: "a good   example"  
输出: "example good a"     
解释: 如果两个单词间有多余的空格，将反转后单词间的空格减少到只含一个。  

```python
def reverseWords_inplace(s):
    def _swap(s, i, j):
        s[i] = chr(ord(s[i]) ^ ord(s[j]))
        s[j] = chr(ord(s[j]) ^ ord(s[i]))
        s[i] = chr(ord(s[i]) ^ ord(s[j]))
    def _reverse(s, i, j):
        while i < j:
            _swap(s, i, j)
            i += 1
            j -= 1
    
    chars = list(s)
    l, r = 0, len(chars)-1
    while l < r:
        _swap(chars, l, r)
        l += 1
        r -= 1
    # print("\""+''.join(chars)+"\"")
    slow, fast = 0, 1
    while fast < len(chars):
        # print(slow, fast)
        while fast<len(chars) and (chars[fast] != ' ' or (fast>0 and chars[fast] == chars[fast-1])):
            fast += 1
        
        r = fast-1
        _reverse(chars, slow, r)
        while slow < len(chars) and chars[slow] != ' ':
            slow += 1
        slow += 1
        fast += 1
    # remove empty
    r = len(chars)
    while chars[r-1] == ' ':
        r -= 1
    return ''.join(chars[:r])
```

```python
def reverseWords_py(s):
    words = [w for w in s.strip().split(' ') if w!='']
    return ' '.join(words[::-1])

def reverseWords(s):
    return reverseWords_inplace(s)
    # return reverseWords_py(s)
```

```python
s = "the sky is blue"
reverseWords(s)
```

```python
s = "  hello world  "
reverseWords(s)
```
