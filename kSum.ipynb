# 1. 两数之和

[力扣题目链接](https://leetcode.cn/problems/two-sum/)

给定一个整数数组 nums 和一个目标值 target，请你在该数组中找出和为目标值的那 两个 整数，并返回他们的数组下标。

你可以假设每种输入只会对应一个答案。但是，数组中同一个元素不能使用两遍。

**示例:**

给定 nums = [2, 7, 11, 15], target = 9

因为 nums[0] + nums[1] = 2 + 7 = 9

所以返回 [0, 1]


```python
def twoSum(nums, target):
    m = {}
    for i,x in enumerate(nums):
        if m.get(target-x) is not None:
            return [m[target-x], i]
        m[x] = i
    return []
```

```python
twoSum([2, 7, 11, 15], 9)
```

```python
twoSum([3,2,4], 6)
```

```python
twoSum([3,3], 6)
```

# 第15题. 三数之和

[力扣题目链接](https://leetcode.cn/problems/3sum/)

给你一个包含 n 个整数的数组 nums，判断 nums 中是否存在三个元素 a，b，c ，使得 a + b + c = 0 ？请你找出所有满足条件且不重复的三元组。

**注意：** 答案中不可以包含重复的三元组。

示例：

给定数组 nums = [-1, 0, 1, 2, -1, -4]，

满足要求的三元组集合为：
[
  [-1, 0, 1],
  [-1, -1, 2]
]

```python
## 这题与two sum不同要求返回去重的value的triplets，也不是索引的triplets
## Hash解 结果的去重很麻烦
def threeSum_hash(nums):
    m = {}
    res = []
    for i,x in enumerate(nums):
        for j,y in enumerate(nums[i+1:]):
            t = 0 - x - y
            if y in m:
                res += [p + [y] for p in m[y]]
            m[t] = m.get(t, [])
            m[t].append([x, y])
            # print((i, j+i+1), (x, y), t, m)
    return res
```

```python
nums = [-1, 0, 1, 2, -1, -4]
triplets = threeSum_hash(nums)
for tri in triplets:
    print(tri, sum(tri))
```

```python
## 双指针解 O(N**2)
def threeSum(nums):
    nums = sorted(nums)
    # print(nums)
    res = []
    for i,x in enumerate(nums):
        if x > 0:
            return res
        ## 跳过相同的数避免重复
        if i>0 and nums[i] == nums[i-1]:
            continue
        l, r = i+1, len(nums) - 1
        while l < r:
            s = x + nums[l] + nums[r]
            if s < 0 or (l>i+1 and nums[l]==nums[l-1]):
                ## 跳过相同的数，只取用最后一个
                l += 1
            elif s > 0 or (r<len(nums)-1 and nums[r]==nums[r+1]):
                ## 跳过相同的数，只取用最前一个
                r -= 1
            else:
                res.append([x, nums[l], nums[r]])
                ## 为什么这里l和r都可以动，比如只做l+=1，则和一定>t了，下一轮r要-1 anyway
                l += 1
                r -= 1
    return res
```

```python
nums = [-1, 0, 1, 2, -1, -4]
triplets = threeSum(nums)
for tri in triplets:
    print(tri, sum(tri))
```

```python
nums = [0, 1, 1]
triplets = threeSum(nums)
for tri in triplets:
    print(tri, sum(tri))
```

```python
nums = [0, 0, 0]
triplets = threeSum(nums)
for tri in triplets:
    print(tri, sum(tri))
```

```python
nums = [1,2,0,1,0,0,0,0]
triplets = threeSum(nums)
for tri in triplets:
    print(tri, sum(tri))
```

# 第18题. 四数之和

[力扣题目链接](https://leetcode.cn/problems/4sum/)

题意：给定一个包含 n 个整数的数组 nums 和一个目标值 target，判断 nums 中是否存在四个元素 a，b，c 和 d ，使得 a + b + c + d 的值与 target 相等？找出所有满足条件且不重复的四元组。

**注意：**

答案中不可以包含重复的四元组。

示例：
给定数组 nums = [1, 0, -1, 0, -2, 2]，和 target = 0。
满足要求的四元组集合为：
[
  [-1,  0, 0, 1],
  [-2, -1, 1, 2],
  [-2,  0, 0, 2]
]

```python
def fourSum(nums, target):
    # return fourSum_doable(nums, target)
    return fourSum_graceful(nums, target)

def fourSum_doable(nums, target):
    nums.sort()
    res = []
    for i,x in enumerate(nums):
        if i>0 and nums[i] == nums[i-1]:
            continue
        j = i+1 ## 不写成for j,y in enumerate(nums[i+1:]) 是因为j会从0开始，需要重新调整成i+1
        while j < len(nums):
            if j>i+1 and nums[j] == nums[j-1]:
                j += 1
                continue
            y = nums[j]
            l, r = j+1, len(nums)-1
            while l < r:
                s = x + y + nums[l] + nums[r]
                if s < target or (l>j+1 and nums[l]==nums[l-1]):
                    l += 1
                elif s > target or (r<len(nums)-1 and nums[r]==nums[r+1]):
                    r -= 1
                else:
                    res.append([x,y,nums[l],nums[r]])
                    l += 1
                    r -= 1
            j += 1
    return res

## 有没有更优雅的写法？
def fourSum_graceful(nums, target):
    def twoSum(nums, target):
        ## note that nums is sorted
        res = []
        l, r = 0, len(nums)-1
        while l < r:
            s = nums[l] + nums[r]
            if s < target or (l>0 and nums[l]==nums[l-1]):
                l += 1
            elif s > target or (r<len(nums)-1 and nums[r]==nums[r+1]):
                r -= 1
            else:
                res.append([nums[l], nums[r]])
                l += 1
                r -= 1
        return res

    def kSum(nums, target, k):
        if k == 2:
            return twoSum(nums, target)
        res = []
        for i in range(len(nums)):
            if i > 0 and nums[i] == nums[i-1]:
                continue
            rs = kSum(nums[i+1:], target-nums[i], k-1)
            res += [[nums[i]]+r for r in rs]
        return res
    
    nums.sort()
    return kSum(nums, target, 4)
```

```python
nums, target = [1,0,-1,0,-2,2], 0
quadrulets = fourSum(nums, target)
for quad in quadrulets:
    print(quad, sum(quad))
```

```python
nums, target = [2,2,2,2,2], 8
quadrulets = fourSum(nums, target)
for quad in quadrulets:
    print(quad, sum(quad))
```

```python
nums, target = [-2,-1,-1,1,1,2,2], 0
quadrulets = fourSum(nums, target)
for quad in quadrulets:
    print(quad, sum(quad))
```

# 第454题.四数相加II

[力扣题目链接](https://leetcode.cn/problems/4sum-ii/)

给定四个包含整数的数组列表 A , B , C , D ,计算有多少个元组 (i, j, k, l) ，使得 A[i] + B[j] + C[k] + D[l] = 0。

为了使问题简单化，所有的 A, B, C, D 具有相同的长度 N，且 0 ≤ N ≤ 500 。所有整数的范围在 -2^28 到 2^28 - 1 之间，最终结果不会超过 2^31 - 1 。

**例如:**

输入:
* A = [ 1, 2]
* B = [-2,-1]
* C = [-1, 2]
* D = [ 0, 2]

输出:

2

**解释:**

两个元组如下:

1. (0, 0, 0, 1) -> A[0] + B[0] + C[0] + D[1] = 1 + (-2) + (-1) + 2 = 0
2. (1, 1, 0, 0) -> A[1] + B[1] + C[0] + D[0] = 2 + (-1) + (-1) + 0 = 0


```python
## 注意到这道题又不一样，首先数组是4个，其次要返回的是个数，也没要求按value去重
## 暴力解是O(N**4)，还是很高的
def fourSumCount(nums1, nums2, nums3, nums4):
    ## O(N**2)
    h = {}
    for x in nums1:
        for y in nums2:
            t = 0 - x - y
            h[t] = h.get(t, 0) + 1
    cnt = 0
    for x in nums3:
        for y in nums4:
            t = x + y
            if t in h:
                cnt += h[t]
    return cnt
```

```python
nums1, nums2, nums3, nums4 = [1,2], [-2,-1], [-1,2], [0,2]
fourSumCount(nums1, nums2, nums3, nums4)
```

```python
nums1, nums2, nums3, nums4 = [0], [0], [0], [0]
fourSumCount(nums1, nums2, nums3, nums4)
```
