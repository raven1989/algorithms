# 24. 两两交换链表中的节点

[力扣题目链接](https://leetcode.cn/problems/swap-nodes-in-pairs/)

给定一个链表，两两交换其中相邻的节点，并返回交换后的链表。

你不能只是单纯的改变节点内部的值，而是需要实际的进行节点交换。

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def to_array(self):
        arr = []
        curr = self
        while curr:
            arr.append(curr.val)
            curr = curr.next
        return arr

def swapPairs(head):
    # return swapPairs_loop(head)
    return swapPairs_recursive(head, 0)

def swapPairs_loop(head):
    handle = ListNode(-1, head)
    curr = handle
    i = 0
    while curr.next and curr.next.next:
        if i % 2 == 0:
            mid = curr.next
            ## curr -> mid -> p2 -> rest
            curr.next = mid.next
            ## curr -> p2 -> rest; mid
            mid.next = curr.next.next
            ## mid -> rest; curr -> p2
            curr.next.next = mid 
            ## curr -> p2 -> mid -> rest
        curr = curr.next
        i += 1
    head = handle.next
    del handle
    return head

def swapPairs_recursive(head, i):
    if not head:
        return head
    curr = swapPairs_recursive(head.next, i+1)
    if i % 2 == 0 and curr:
        head.next = curr.next
        curr.next = head
        return curr
    else:
        head.next = curr
        return head
```

```python
def make_list(arr):
    arr.reverse()
    head = None
    for x in arr:
        head = ListNode(x, head)
    return head
```

```python
head = make_list([1,2,3,4])
print(head.to_array() if head else [])
head = swapPairs(head)
print(head.to_array() if head else [])
```
