# 343. 整数拆分

[力扣题目链接](https://leetcode.cn/problems/integer-break/)

给定一个正整数 n，将其拆分为至少两个正整数的和，并使这些整数的乘积最大化。 返回你可以获得的最大乘积。

示例 1:
* 输入: 2
* 输出: 1
* 解释: 2 = 1 + 1, 1 × 1 = 1。

示例 2:
* 输入: 10
* 输出: 36
* 解释: 10 = 3 + 3 + 4, 3 × 3 × 4 = 36。
* 说明: 你可以假设 n 不小于 2 且不大于 58。


```python
def integerBreak(n):
    dp = [1] * (n+1)
    dp[:3] = [1,1,1]
    for i in range(3, n+1):
        for j in range(1, i//2+1):
            # dp[i] = max(max(dp[j],j) * max(dp[i-j],i-j), dp[i])
            ## j 为什么可以不考虑拆的情况，考虑 7
            ## 1 + 6, 2 + 5, 3 + 4中，
            ## 再考虑3 + 4: 如果拆3，对应组合 1 + 2 + 4
            ## 再考虑2 + 5: 不拆2，拆5有考虑过 2 + 1 + 4
            ## 所以，可以只拆后面那个数
            # print(i, i//2, j, i-j)
            dp[i] = max(j * max(dp[i-j],i-j), dp[i])
    return dp[-1]
```
