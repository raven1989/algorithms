# 206.反转链表

[力扣题目链接](https://leetcode.cn/problems/reverse-linked-list/)

题意：反转一个单链表。

示例:
输入: 1->2->3->4->5->NULL
输出: 5->4->3->2->1->NULL

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    def to_array(self):
        arr = []
        curr = self
        while curr:
            arr.append(curr.val)
            curr = curr.next
        return arr

def reverseList_loop(head):
    prev, curr = None, head
    while curr:
        tmp = curr.next
        curr.next = prev
        prev = curr
        curr = tmp
    return prev

def reverseList_recursive(head):
    if not head:
        return head
    if not head.next:
        return head
    last = reverseList_recursive(head.next)
    head.next.next = head
    head.next = None
    return last

def reverseList(head):
    # return reverseList_loop(head)
    return reverseList_recursive(head)
```

```python
def make_list(arr):
    arr.reverse()
    head = None
    for x in arr:
        head = ListNode(x, head)
    return head
```

```python
head = make_list([1,2,3,4,5])
print(head.to_array() if head else [])
head = reverseList(head)
print(head.to_array() if head else [])
```

```python
head = make_list([1,2])
print(head.to_array() if head else [])
head = reverseList(head)
print(head.to_array() if head else [])
```

```python
head = make_list([])
print(head.to_array() if head else [])
head = reverseList(head)
print(head.to_array() if head else [])
```
