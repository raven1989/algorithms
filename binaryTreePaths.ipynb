# 257. 二叉树的所有路径

[力扣题目链接](https://leetcode.cn/problems/binary-tree-paths/)

给定一个二叉树，返回所有从根节点到叶子节点的路径。

说明: 叶子节点是指没有子节点的节点。

示例:
![257.二叉树的所有路径1](https://file1.kamacoder.com/i/algo/2021020415161576.png)


```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Tree:
    @staticmethod
    def build(arr):
        q = []
        i = j = 0
        while j < len(arr):
            if len(q) == 0:
                q.append(TreeNode(arr[j]))
                j += 1
            else:
                q[i].left = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                q[i].right = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                if q[i].left:
                    q.append(q[i].left)
                if q[i].right:
                    q.append(q[i].right)
                i += 1
        # print([it.val if it else it for it in q])
        return q[0] if len(q)>0 else None
    
    @staticmethod
    def print(root):
        ## 如何print一棵树，如果左和右子树都画好了，就能画根了
        def _postorder_print(r):
            if not r:
                return ['']
            left = _postorder_print(r.left)
            right = _postorder_print(r.right)
            res = ['']
            lines = max(len(left), len(right))
            for i in range(lines):
                ll = left[i] if i<len(left) else ' '*len(left[0])
                rr = right[i] if i<len(right) else ' '*len(right[0])
                res.append([ll, ' '*len(str(r.val)), rr])
            res[0] = [
                '' if len(res)<2 else ' '*len(res[1][0]),
                str(r.val),
                '' if len(res)<2 else ' '*len(res[1][2])
            ]
            res = [''.join(line) for line in res]
            # print(res)
            return res
        return '\n'.join([line for line in _postorder_print(root) if line.strip()!=''])
```

```python
def binaryTreePaths(root):
    if not root:
        return []
    if not root.left and not root.right:
        return [str(root.val)]
    left_paths = binaryTreePaths(root.left)
    right_paths = binaryTreePaths(root.right)
    return [str(root.val)+"->"+p for p in left_paths+right_paths]
```

```python
root = Tree.build([1,2,3,None,5])
print("Tree:")
print(Tree.print(root))
binaryTreePaths(root)
```

```python
root = Tree.build([1])
print("Tree:")
print(Tree.print(root))
binaryTreePaths(root)
```
