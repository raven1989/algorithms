# 第202题. 快乐数

[力扣题目链接](https://leetcode.cn/problems/happy-number/)

编写一个算法来判断一个数 n 是不是快乐数。

「快乐数」定义为：对于一个正整数，每一次将该数替换为它每个位置上的数字的平方和，然后重复这个过程直到这个数变为 1，也可能是 无限循环 但始终变不到 1。如果 可以变为  1，那么这个数就是快乐数。

如果 n 是快乐数就返回 True ；不是，则返回 False 。

**示例：**

输入：19     
输出：true    
解释：           
1^2 + 9^2 = 82        
8^2 + 2^2 = 68     
6^2 + 8^2 = 100   
1^2 + 0^2 + 0^2 = 1    

```python
def isHappy(n):
    # Not Happy Number : 每位平方和会出现重复数字
    m = {}
    d, sum = n, 0

    while True:
        # print(d)
        while d > 0:
            sum += (d % 10) ** 2
            d = d // 10
        if sum in m:
            return False
        if sum == 1:
            return True
        m[sum] = 1
        d = sum
        sum = 0
```

```python
isHappy(19)
```

```python
isHappy(2)
```
