# 二叉树的深度遍历

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Tree:
    @staticmethod
    def build(arr):
        q = []
        i = j = 0
        while j < len(arr):
            if len(q) == 0:
                q.append(TreeNode(arr[j]))
                j += 1
            else:
                q[i].left = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                q[i].right = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                if q[i].left:
                    q.append(q[i].left)
                if q[i].right:
                    q.append(q[i].right)
                i += 1
        # print([it.val if it else it for it in q])
        return q[0] if len(q)>0 else None
    
    @staticmethod
    def print(root):
        ## 如何print一棵树，如果左和右子树都画好了，就能画根了
        def _postorder_print(r):
            if not r:
                return ['']
            left = _postorder_print(r.left)
            right = _postorder_print(r.right)
            res = ['']
            lines = max(len(left), len(right))
            for i in range(lines):
                ll = left[i] if i<len(left) else ' '*len(left[0])
                rr = right[i] if i<len(right) else ' '*len(right[0])
                res.append([ll, ' '*len(str(r.val)), rr])
            res[0] = [
                '' if len(res)<2 else ' '*len(res[1][0]),
                str(r.val),
                '' if len(res)<2 else ' '*len(res[1][2])
            ]
            res = [''.join(line) for line in res]
            # print(res)
            return res
        return '\n'.join([line for line in _postorder_print(root) if line.strip()!=''])
```

## 144.二叉树的前序遍历

```python
def preorderTraversal(root):
    # return preorderTraversal_recursive(root)
    return preorderTraversal_loop(root)

def preorderTraversal_recursive(root):
    res = []
    if root:
        res += [root.val]
        res += preorderTraversal(root.left)
        res += preorderTraversal(root.right)
    return res

def preorderTraversal_loop(root):
    visit = []
    stack = []
    curr = root
    while len(stack) > 0 or curr:
        if curr:
            visit.append(curr.val)
            stack.append(curr)
            curr = curr.left
        else:
            curr = stack.pop()
            curr = curr.right
    return visit
```

## 94.二叉树的中序遍历

```python
def inorderTraversal(root):
    # return inorderTraversal_recursive(root)
    return inorderTraversal_loop(root)

def inorderTraversal_recursive(root):
    res = []
    if root:
        res += inorderTraversal(root.left)
        res += [root.val]
        res += inorderTraversal(root.right)
    return res

def inorderTraversal_loop(root):
    visit = []
    stack = []
    curr = root
    while len(stack) > 0 or curr:
        if curr:
            stack.append(curr)
            curr = curr.left
        else:
            curr = stack.pop()
            visit.append(curr.val)
            curr = curr.right
    return visit
```

## 145.二叉树的后序遍历

```python
def postorderTraversal(root):
    # return postorderTraversal_recursive(root)
    return postorderTraversal_loop(root)

def postorderTraversal_recursive(root):
    res = []
    if root:
        res += postorderTraversal(root.left)
        res += postorderTraversal(root.right)
        res += [root.val]
    return res

def postorderTraversal_loop(root):
    visit = []
    stack = []
    ## (node, bool) bool标记是否右子已经压栈，如果已经压栈，下次访问到时证明其左右子均已访问过，表示可以访问
    curr = root
    while len(stack) > 0 or curr:
        if curr:
            stack.append((curr, False))
            curr = curr.left
        else:
            curr, visitable = stack.pop()
            if visitable:
                visit.append(curr.val)
                ## curr 左右子都访问过了，因此=None，不能=stack.pop()，否则又会把stack[-1]的左子树再压栈
                curr = None
            else:
                stack.append((curr, True))
                curr = curr.right
    return visit
```

```python
root = Tree.build([1,None,2,3])
print("Tree:")
print(Tree.print(root))
print("preorder: ", preorderTraversal_recursive(root), preorderTraversal_loop(root))
print("inorder: ", inorderTraversal_recursive(root), inorderTraversal_loop(root))
print("postorder: ", postorderTraversal_recursive(root), postorderTraversal_loop(root))
```

# 二叉树的层序遍历

## 107 二叉树的层序遍历II

Given the root of a binary tree, return the bottom-up level order traversal of its nodes' values. (i.e., from left to right, level by level from leaf to root).

Example 1:


Input: root = [3,9,20,null,null,15,7]
Output: [[15,7],[9,20],[3]]
Example 2:

Input: root = [1]
Output: [[1]]
Example 3:

Input: root = []
Output: []
 

Constraints:

The number of nodes in the tree is in the range [0, 2000].
-1000 <= Node.val <= 1000

```python
def levelOrderBottom(root):
    if not root:
        return []
    visit = []
    q = [root]
    while len(q) > 0:
        next_q = []
        visitable = []
        for n in q:
            visitable.append(n.val)
            if n.left:
                next_q.append(n.left)
            if n.right:
                next_q.append(n.right)
        visit.append(visitable)
        q = next_q
    return visit[::-1]
```

```python
root = Tree.build([3,9,20,None,None,15,7])
print("Tree:")
print(Tree.print(root))
levelOrderBottom(root)
```

```python
root = Tree.build([1])
print("Tree:")
print(Tree.print(root))
levelOrderBottom(root)
```

```python
root = Tree.build([])
print("Tree:")
print(Tree.print(root))
levelOrderBottom(root)
```
