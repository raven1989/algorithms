# 199.二叉树的右视图

给定一个二叉树的 根节点 root，想象自己站在它的右侧，按照从顶部到底部的顺序，返回从右侧所能看到的节点值。

示例 1：

输入：root = [1,2,3,null,5,null,4]

输出：[1,3,4]

解释：

示例 2：

输入：root = [1,2,3,4,null,null,null,5]

输出：[1,3,4,5]

解释：

示例 3：

输入：root = [1,null,3]

输出：[1,3]

示例 4：

输入：root = []

输出：[]

提示:

二叉树的节点个数的范围是 [0,100]
-100 <= Node.val <= 100 

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Tree:
    @staticmethod
    def build(arr):
        q = []
        i = j = 0
        while j < len(arr):
            if len(q) == 0:
                q.append(TreeNode(arr[j]))
                j += 1
            else:
                q[i].left = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                q[i].right = TreeNode(arr[j]) if j<len(arr) and arr[j] else None
                j += 1
                if q[i].left:
                    q.append(q[i].left)
                if q[i].right:
                    q.append(q[i].right)
                i += 1
        # print([it.val if it else it for it in q])
        return q[0] if len(q)>0 else None
    
    @staticmethod
    def print(root):
        ## 如何print一棵树，如果左和右子树都画好了，就能画根了
        def _postorder_print(r):
            if not r:
                return ['']
            left = _postorder_print(r.left)
            right = _postorder_print(r.right)
            res = ['']
            lines = max(len(left), len(right))
            for i in range(lines):
                ll = left[i] if i<len(left) else ' '*len(left[0])
                rr = right[i] if i<len(right) else ' '*len(right[0])
                res.append([ll, ' '*len(str(r.val)), rr])
            res[0] = [
                '' if len(res)<2 else ' '*len(res[1][0]),
                str(r.val),
                '' if len(res)<2 else ' '*len(res[1][2])
            ]
            res = [''.join(line) for line in res]
            # print(res)
            return res
        return '\n'.join([line for line in _postorder_print(root) if line.strip()!=''])
```

```python
def rightSideView(root):
    if not root:
        return []
    visit = []
    q = [root]
    while len(q) > 0:
        visit.append(q[-1].val)
        next_level = []
        for n in q:
            if n.left:
                next_level.append(n.left)
            if n.right:
                next_level.append(n.right)
            q = next_level
    return visit
```

```python
root = Tree.build([1,2,3,None,5,None,4])
print("Tree:")
print(Tree.print(root))
rightSideView(root)
```

```python
root = Tree.build([1,2,3,4,None,None,None,5])
print("Tree:")
print(Tree.print(root))
rightSideView(root)
```

```python
root = Tree.build([1,None,3])
print("Tree:")
print(Tree.print(root))
rightSideView(root)
```

```python
root = Tree.build([])
print("Tree:")
print(Tree.print(root))
rightSideView(root)
```
