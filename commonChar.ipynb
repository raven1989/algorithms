# 1002. 查找常用字符

[力扣题目链接](https://leetcode.cn/problems/find-common-characters/)

给你一个字符串数组 words ，请你找出所有在 words 的每个字符串中都出现的共用字符（ 包括重复字符），并以数组形式返回。你可以按 任意顺序 返回答案。

示例 1：

输入：words = ["bella","label","roller"]
输出：["e","l","l"]

示例 2：

输入：words = ["cool","lock","cook"]
输出：["c","o"]

提示：

1 <= words.length <= 100
1 <= words[i].length <= 100
words[i] 由小写英文字母组成

```python
def commonChars(words):
    if len(words) <= 0:
        return []
    res = []
    hash = [0] * 26
    for c in words[0]:
        hash[ord(c) - ord('a')] += 1
    for w in words[1:]:
        hashw = [0] * 26
        for c in w:
            hashw[ord(c) - ord('a')] += 1
        for i in range(26):
            hash[i] = min(hash[i], hashw[i])
    for i in range(26):
        while hash[i] > 0:
            res.append(chr(i + ord('a')))
            hash[i] -= 1
    return res
```

```python
words = ["bella","label","roller"]
commonChars(words)
```

```python
words = ["cool","lock","cook"]
commonChars(words)
```

```python
words = ["cool"]
commonChars(words)
```
