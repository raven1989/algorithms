# 216.组合总和III

[力扣题目链接](https://leetcode.cn/problems/combination-sum-iii/)

找出所有相加之和为 n 的 k 个数的组合。组合中只允许含有 1 - 9 的正整数，并且每种组合中不存在重复的数字。

说明：

* 所有数字都是正整数。
* 解集不能包含重复的组合。

示例 1:
输入: k = 3, n = 7
输出: [[1,2,4]]

示例 2:
输入: k = 3, n = 9
输出: [[1,2,6], [1,3,5], [2,3,4]]

````python
def combinationSumIII(k, n):
    res = []
    path = []

    def backtrack(k, n, start, path, res):
        ## trim 1: 如果已经>目标值，不用继续了
        if n < 0:
            return
        if len(path) == k:
            if n == 0:
                res.append([x for x in path])
            return 
        ## trim 2: i没必要增加到剩余可选元素不足k-len(path)个
        for i in range(start, 10 - (k-len(path)) + 1):
            path.append(i)
            backtrack(k, n-i, i+1, path, res)
            path.pop()
    
    backtrack(k, n, 1, path, res)
    return res
````

````python
combinationSumIII(3, 7)
````

````python
combinationSumIII(3, 9)
````

````python
combinationSumIII(4, 1)
````

# 39. 组合总和

[力扣题目链接](https://leetcode.cn/problems/combination-sum/)

给定一个无重复元素的数组 candidates 和一个目标数 target ，找出 candidates 中所有可以使数字和为 target 的组合。

candidates 中的数字可以无限制重复被选取。

说明：

* 所有数字（包括 target）都是正整数。
* 解集不能包含重复的组合。

示例 1：

* 输入：candidates = [2,3,6,7], target = 7,
* 所求解集为：
  [
  [7],
  [2,2,3]
  ]

示例 2：

* 输入：candidates = [2,3,5], target = 8,
* 所求解集为：
  [
  [2,2,2,2],
  [2,3,3],
  [3,5]
  ]

Constraints:

- 1 <= candidates.length <= 30
- 2 <= candidates[i] <= 40
- All elements of candidates are distinct.
- 1 <= target <= 40

````python
def combinationSum(candidates, target):
    res = []
    path = []

    def backtrack(candidates, target, start, path, res):
        if target < 0:
            return
        if target == 0:
            res.append([x for x in path])
            return
        for i in range(start, len(candidates)):
            x = candidates[i]
            path.append(x)
            ## 因为可以重复选择，比如candidates=[2,3,6,7], path=[2]，下一个选择的candidates仍然是[2,3,6,7]，所以其实start不变；
            ## 但为了避免重复组合出现，当path=[3]时，其实是从path=[2]循环到下一个，此时无论在哪个位置上的candidates都是[3,6,7]，所以这里start=i
            backtrack(candidates, target-x, i, path, res)
            path.pop()

    backtrack(candidates, target, 0, path, res)
    return res
````

````python
combinationSum(candidates = [2,3,6,7], target = 7)
````

````python
combinationSum(candidates = [2,3,5], target = 8)
````

````python
combinationSum(candidates = [2], target = 1)
````

# 40.组合总和II

[力扣题目链接](https://leetcode.cn/problems/combination-sum-ii/)

给定一个数组 candidates 和一个目标数 target ，找出 candidates 中所有可以使数字和为 target 的组合。

candidates 中的每个数字在每个组合中只能使用一次。

说明：
所有数字（包括目标数）都是正整数。解集不能包含重复的组合。 

* 示例 1:
* 输入: candidates = [10,1,2,7,6,1,5], target = 8,
* 所求解集为:
```
[
  [1, 7],
  [1, 2, 5],
  [2, 6],
  [1, 1, 6]
]
```

* 示例 2:
* 输入: candidates = [2,5,2,1,2], target = 5,
* 所求解集为:

```
[
  [1,2,2],
  [5]
]
```

````python
def combinationSum2(candidates, target):
    res = []
    path = []

    ## 注意到candidates中有重复元素，为了避免结果中出现重复组合，对candidates排序
    ## 在同一层中如果选取过的元素，下一次循环直接跳过就可以避免，
    ## cadidates=[1,1,7] 会产生两个path=[1,7]，只要选取第一个元素时跳过第二个1就能避免
    ## 同时保住path=[1,1]这样的组合
    candidates.sort()

    def backtrack(candidates, target, start, path, res):
        if target < 0:
            return
        if target == 0:
            res.append([x for x in path])
            return

        for i in range(start, len(candidates)):
            ## 跳过同层identical的元素避免重复组合，注意i>start而不是i>0
            if i > start and candidates[i]==candidates[i-1]:
                continue
            path.append(candidates[i])
            backtrack(candidates, target-candidates[i], i+1, path, res)
            path.pop()

    backtrack(candidates, target, 0, path, res)
    return res
````

````python
combinationSum2(candidates = [10,1,2,7,6,1,5], target = 8)
````

````python
combinationSum2(candidates = [2,5,2,1,2], target = 5)
````

````python
combinationSum2(candidates = [], target = 3)
````

# 377. 组合总和 Ⅳ

[力扣题目链接](https://leetcode.cn/problems/combination-sum-iv/)

难度：中等

给定一个由正整数组成且不存在重复数字的数组，找出和为给定目标正整数的组合的个数。

示例:

* nums = [1, 2, 3]
* target = 4

所有可能的组合为：
(1, 1, 1, 1)
(1, 1, 2)
(1, 2, 1)
(1, 3)
(2, 1, 1)
(2, 2)
(3, 1)

请注意，顺序不同的序列被视作不同的组合。

因此输出为 7。

````python
def combinationSum4(nums, target):
    ## 顺序不同被视作不同的结果，这意味着：
    ## 以[1,2,3]为例，第一个元素可从1,2,3中选择，第二个元素还可以从1,2,3中选择
    ## 比如：1211这条路径，2111这条路径都要保留；
    ## 如果要求组合，即1211和2111被视作相同的话，第一个元素选择了2之后，接下来只能从2,3中选择了
    ## 因此，只统计数量的话，用dp就够了，不需要回溯，是个完全背包问题
    ## dp[i][j]就表示从0-i中选择，和为j的排列数
    ## 遍历顺序决定是组合还是排列：
    ## 如果先遍历物品再背包容量就是组合：因为nums本身是有顺序的，不会出现21这个顺序的组合
    ## 如果先遍历背包容量再物品就是排列：思考下含义是背包容量为j的情况下去遍历nums的0-i，所以是排列
    dp = [0] * (target + 1)
    dp[0] = 1

    ## 如果不选nums[i]，和为j的排列数为dp[i-1][j]
    ## 如果当前这一个选i，和为j的排列数为dp[i][j-nums[i]]
    for j in range(target+1):
        for i in range(len(nums)):
            if j >= nums[i]:
                dp[j] += dp[j-nums[i]]
        # print(dp)
    return dp[-1]
````

````python
nums = [1,2,3]
target = 4
combinationSum4(nums, target)
````

````python
nums = [9]
target = 3
combinationSum4(nums, target)
````
