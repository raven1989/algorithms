# 242.有效的字母异位词

[力扣题目链接](https://leetcode.cn/problems/valid-anagram/)

给定两个字符串 s 和 t ，编写一个函数来判断 t 是否是 s 的字母异位词。

示例 1:
输入: s = "anagram", t = "nagaram"
输出: true

示例 2:
输入: s = "rat", t = "car"
输出: false

**说明:**
你可以假设字符串只包含小写字母。

```python
def isAnagram(s, t):
    if len(s) <= 0 or len(t) <=0:
        return False
    hash = [0] * 26
    for x in s:
        hash[ord(x) - ord('a')] += 1
    for x in t:
        hash[ord(x) - ord('a')] -= 1
    for x in hash:
        if x != 0:
            return False
    return True
```

```python
s, t = "anagram", "nagaram"
print(isAnagram(s, t))
```

```python
s, t = "rat", "car"
print(isAnagram(s, t))
```

```python
s, t = "a", "ab"
print(isAnagram(s, t))
```

```python
s, t = "ab", "a"
print(isAnagram(s, t))
```
